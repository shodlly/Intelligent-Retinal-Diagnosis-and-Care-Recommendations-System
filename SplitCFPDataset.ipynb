```python
import splitfolders
import os

input_folder = "/content/ocular-diseases-asma/Eye-diseasesasma"
splitfolders.ratio(input_folder, output="dataset_split01", seed=42, ratio=(0.7, 0.0, 0.3))
splitfolders.ratio("dataset_split/train", output="final_dataset01", seed=42, ratio=(0.8, 0.2))

```

```python
def count_images(directory):
    if not os.path.exists(directory):
        return f"Error: '{directory}' does not exist."

    class_counts = {}
    for class_name in os.listdir(directory):
        class_path = os.path.join(directory, class_name)
        if os.path.isdir(class_path):
            class_counts[class_name] = len(os.listdir(class_path))
    return class_counts

print("Training set:", count_images("final_dataset01/train"))
print("Validation set:", count_images("final_dataset01/val"))
print("Testing set:", count_images("dataset_split01/test"))
```

```python
import os
import shutil


train_src = "final_dataset01/train"
val_src = "final_dataset01/val"
test_src = "dataset_split01/test"


output_folder = "complete_dataset"
os.makedirs(output_folder, exist_ok=True)


train_dest = os.path.join(output_folder, "train")
val_dest = os.path.join(output_folder, "val")
test_dest = os.path.join(output_folder, "test")

os.makedirs(train_dest, exist_ok=True)
os.makedirs(val_dest, exist_ok=True)
os.makedirs(test_dest, exist_ok=True)

# Function to move files
def move_files(src, dest):
    for class_name in os.listdir(src):
        src_class_path = os.path.join(src, class_name)
        dest_class_path = os.path.join(dest, class_name)

        if os.path.isdir(src_class_path):
            os.makedirs(dest_class_path, exist_ok=True)
            for file in os.listdir(src_class_path):
                shutil.move(os.path.join(src_class_path, file), os.path.join(dest_class_path, file))

# Move files to the new structure
move_files(train_src, train_dest)
move_files(val_src, val_dest)
move_files(test_src, test_dest)

print("Files successfully moved to 'complete_dataset' with 'train', 'val', and 'test' folders.")
```
